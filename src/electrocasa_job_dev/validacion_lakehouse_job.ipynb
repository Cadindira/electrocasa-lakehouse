```python
# ============================================================
# VALIDACIÓN AUTOMÁTICA DEL LAKEHOUSE ELECTROCASA
# ============================================================
#
# Este notebook se ejecutará después del pipeline principal.
# Si alguna validación crítica no se cumple, se genera un error
# y el Job queda marcado como fallido.
# ============================================================


def contar(tabla):
    return spark.table(tabla).count()


validaciones = {
    "ventas_silver": (
        contar("electrocasa.silver.ventas"),
        13886
    ),

    "ventas_cuarentena": (
        contar("electrocasa.silver.ventas_cuarentena"),
        1114
    ),

    "productos_silver": (
        contar("electrocasa.silver.productos"),
        2849
    ),

    "productos_cuarentena": (
        contar("electrocasa.silver.productos_cuarentena"),
        181
    ),

    "tracking_silver": (
        contar("electrocasa.silver.tracking_envios"),
        5000
    ),

    "empleados_silver": (
        contar("electrocasa.silver.empleados"),
        4032
    ),

    "empleados_cuarentena": (
        contar("electrocasa.silver.empleados_cuarentena"),
        46
    ),

    "empleados_scd2": (
        contar("electrocasa.gold.empleados_scd2"),
        4017
    ),

    "devoluciones_silver": (
        contar("electrocasa.silver.devoluciones"),
        3892
    ),

    "devoluciones_cuarentena": (
        contar("electrocasa.silver.devoluciones_cuarentena"),
        108
    ),

    "resenas_silver": (
        contar("electrocasa.silver.resenas"),
        7705
    ),

    "resenas_cuarentena": (
        contar("electrocasa.silver.resenas_cuarentena"),
        295
    )
}


errores = []

for nombre, (valor_real, valor_esperado) in validaciones.items():

    if valor_real == valor_esperado:
        print(
            f"OK - {nombre}: "
            f"{valor_real}"
        )
    else:
        mensaje = (
            f"ERROR - {nombre}: "
            f"esperado={valor_esperado}, "
            f"real={valor_real}"
        )

        print(mensaje)
        errores.append(mensaje)


if errores:
    raise Exception(
        "Fallaron validaciones del Lakehouse:\n"
        + "\n".join(errores)
    )


print("Todas las validaciones del Lakehouse finalizaron correctamente.")
```
